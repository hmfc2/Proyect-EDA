```python
#Como se ve los datos de ACT, que son los de actividades a nivel nacional, pero los datos sacados de SERNATUR, vienen en un 
# excel, de distintas hojas, se busco la hoja de las actividades a nivel comunal y los valores en UF, y se guardo de forma .csv 
# link de acceso: https://drive.google.com/file/d/1DLxzQz7CeGgUT6wWUt3fNiYqu9lshYBc/view?usp=sharing

#

file_id_encuesta = "1DLxzQz7CeGgUT6wWUt3fNiYqu9lshYBc"

url = f"https://drive.google.com/uc?id={file_id_encuesta}"

ventas = pd.read_csv(
    url,
    sep=";",
    engine="python"
)

ventas.head(20)
```

```python
# Como se ven los datos no sirven para ser ocupados, ya que las columnas son las columnas, y en las filas se encuentran las 
# comunas y en suborden las actividades. Entonces, se va a realizar una tabla con solo los datos de San fernando. 
# Primero se va mirar como va quedar la tabla son con las columnas que se van a ocupar. 
print(ventas.columns.tolist())
ventas_2023 = ventas[['Región / CUT Comuna ', ' Nombre Comuna / ACT', '2023']]
ventas_2023.head(20)
```

```python
#Luego se va a realizar la tabla solo de la comuna de San Fernando.
sf=ventas[ventas.iloc[:,1].astype(str).str.contains("San Fernando", case=False, na=False)]
indice_sf = sf.index[0]

sft = ventas.iloc[indice_sf+1:indice_sf+13, [1,2]]
sft.columns = ["Actividad", "Ventas_2023"]
print(sft)
```
